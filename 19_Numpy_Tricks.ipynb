### np.sort

Return a sorted copy of an array.

https://numpy.org/doc/stable/reference/generated/numpy.sort.html

```python
# code
import numpy as np
a = np.random.randint(1,100,15)
a
```

```python
b = np.random.randint(1,100,24).reshape(6,4)
b
```

```python
# ascending sort
np.sort(a)
```

```python
# descending sort
np.sort(a)[::-1]
```

```python
# col-wise sort
np.sort(b,axis=1)
np.sort(b,axis=0)
```

```python
# row-wise sort
np.sort(b,axis=1)
```

### Advanced NumPy Functions

```python
# np.swapaxes
# Swap the axes of a 2D array b
print("Original array b:\n", b)
swapped_b = np.swapaxes(b, 0, 1)
print("\nArray b after swapping axes:\n", swapped_b)
```

```python
# np.random.uniform (assuming 'np.uniform' refers to this)
# Generate 5 random numbers between 0 and 1 from a uniform distribution
uniform_numbers = np.random.uniform(low=0.0, high=1.0, size=5)
print("Uniformly distributed random numbers:\n", uniform_numbers)
```

```python
# np.count_nonzero
# Count non-zero elements in array 'a'
print("Original array a:\n", a)
non_zero_count_a = np.count_nonzero(a)
print("Number of non-zero elements in 'a':", non_zero_count_a)

# Create an array with zeros to demonstrate
arr_with_zeros = np.array([0, 1, 0, 5, 0, 3])
non_zero_count_zeros = np.count_nonzero(arr_with_zeros)
print("Array with zeros:\n", arr_with_zeros)
print("Number of non-zero elements in arr_with_zeros:", non_zero_count_zeros)
```

```python
# np.tile
# https://www.kaggle.com/code/abhayparashar31/best-numpy-functions-for-data-science-50?scriptVersionId=98816580

x = np.array([0, 1, 2])
print("Original array x:\n", x)

# Tile x 3 times along the first dimension
tiled_x = np.tile(x, 3)
print("\nArray tiled 3 times:\n", tiled_x)

# Tile x with specific repetitions
y = np.array([[0, 1], [2, 3]])
print("\nOriginal array y:\n", y)
tiled_y = np.tile(y, (2, 2))
print("\nArray y tiled (2,2) times:\n", tiled_y)
```

```python
# np.repeat
# https://towardsdatascience.com/10-numpy-functions-you-should-know-1dc4863764c5

x = np.array([1, 2, 3])
print("Original array x:\n", x)

# Repeat each element 2 times
repeated_x = np.repeat(x, 2)
print("\nElements repeated 2 times:\n", repeated_x)

# Repeat with a 2D array along axis 0
y = np.array([[1, 2], [3, 4]])
print("\nOriginal array y:\n", y)
repeated_y_axis0 = np.repeat(y, 2, axis=0)
print("\nElements repeated along axis 0:\n", repeated_y_axis0)

# Repeat with a 2D array along axis 1
repeated_y_axis1 = np.repeat(y, 2, axis=1)
print("\nElements repeated along axis 1:\n", repeated_y_axis1)
```

```python
# np.allclose and np.array_equal (for 'equals')

# np.allclose: check if two arrays are element-wise equal within a tolerance
a1 = np.array([1.0, 2.0, 3.0])
a2 = np.array([1.00000001, 2.0, 2.99999999])
a3 = np.array([1.1, 2.0, 3.0])

print(f"Are a1 and a2 allclose? {np.allclose(a1, a2)}")
print(f"Are a1 and a3 allclose? {np.allclose(a1, a3)}")
print(f"Are a1 and a3 allclose with custom tolerance? {np.allclose(a1, a3, atol=0.1)}")

# np.array_equal: check if two arrays have the same shape and elements
b1 = np.array([1, 2, 3])
b2 = np.array([1, 2, 3])
b3 = np.array([1, 2, 4])
b4 = np.array([[1, 2, 3]])

print(f"\nAre b1 and b2 array_equal? {np.array_equal(b1, b2)}")
print(f"Are b1 and b3 array_equal? {np.array_equal(b1, b3)}")
print(f"Are b1 and b4 array_equal (different shape)? {np.array_equal(b1, b4)}")

# Direct equality check (element-wise comparison)
print(f"\nElement-wise equality (b1 == b2): {b1 == b2}")
print(f"Element-wise equality (b1 == b3): {b1 == b3}")
```
