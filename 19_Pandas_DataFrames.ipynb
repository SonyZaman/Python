### Creating DataFrame

```python
# using lists
student_data = [
    [100,80,10],
    [90,70,7],
    [120,100,14],
    [80,50,2]
]

pd.DataFrame(student_data,columns=['iq','marks','package'])
```

```python
# using dicts

student_dict = {
    'name':['Sony','Labib','Hasan','Rahman','Zaman','Karim'],
    'iq':[100,90,120,80,0,0],
    'marks':[80,70,100,50,0,0],
    'package':[10,7,14,2,0,0]
}

students = pd.DataFrame(student_dict)
students.set_index('name',inplace=True)
students
```

### Filtering a DataFrame

```python
cricket.head(2)
```

```python
# find all the final winners
mask = cricket['MatchNumber'] == 'Final'
new_df = cricket[mask]
new_df[['Season','WinningTeam']]
```

```python
# in one line
cricket[cricket['MatchNumber'] == 'Final'][['Season','WinningTeam']]
```

```python
# how many super over finishes have occured
cricket[cricket['SuperOver'] == 'Y'].shape[0]
```

```python
# how many matches has Australia won in Sydney
cricket[(cricket['City'] == 'Sydney') & (cricket['WinningTeam'] == 'Australia')].shape[0]
```

```python
# toss winner is match winner in percentage
(cricket[cricket['TossWinner'] == cricket['WinningTeam']].shape[0]/cricket.shape[0])*100
```

```python
# movies with rating higher than 8 and votes>10000
movies[(movies['imdb_rating'] > 8.0) & (movies['imdb_votes'] > 10000)].shape[0]
```

```python
# Action movies with rating higher than 7.5
# mask1 = movies['genres'].str.split('|').apply(lambda x:'Action' in x)
mask1 = movies['genres'].str.contains('Action')
mask2 = movies['imdb_rating'] > 7.5

movies[mask1 & mask2]
```

```python
# write a function that can return the track record of 2 teams against each other
cricket[((cricket['Team1']=='Australia') & (cricket['Team2']=='England') )|((cricket['Team1']=='England') & (cricket['Team2']=='Australia') )]['Team1']
```

### Adding new cols

```python
# add new column Certified
movies['Certified'] = True
movies.head()
```

```python
# Remove rows that contain missing values
movies.dropna(inplace=True)
```

```python
# Add a new column named 'prize_count', where the values are extracted from 'wins_nominations'
movies['prize_count'] = movies['wins_nominations'].str.split(' ').apply(lambda x: x[0])

movies.head()
```

```python
movies.info()
```

### Important DataFrame Functions

```python
# astype() is used to change the data type of a given column
cricket.info()
```

```python
cricket['ID'] = cricket['ID'].astype('int32')
```

```python
cricket.info()
```

```python
# Convert 'toss_decision' and 'super_over' columns into category data type
cricket['TossDecision'] = cricket['TossDecision'].astype('category')
cricket['SuperOver'] = cricket['SuperOver'] .astype('category')
```

```python
cricket.info()
```

```python
cricket['City'].value_counts()
```

```python
# how many matches each team has played
pd.concat([cricket['Team1'], cricket['Team2']]).value_counts()
```

```python
# sort_values -> ascending -> na_position -> inplace -> multiple cols
students.sort_values(by=['iq', 'percent'], ascending=[True, False], na_position='first', inplace=True)
students
```
